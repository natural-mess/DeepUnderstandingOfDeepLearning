```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt

# NEW!
import scipy.stats as stats
```

```python
# parameters
n1 = 30   # samples in dataset 1
n2 = 40   # ...and 2
mu1 = 1   # population mean in dataset 1
mu2 = 2   # population mean in dataset 2

# generate the data
data1 = mu1 + np.random.randn(n1)
data2 = mu2 + np.random.randn(n2)



# plot them
plt.plot(np.zeros(n1),data1,'ro',markerfacecolor='w',markersize=14)
plt.plot(np.ones(n2), data2,'bs',markerfacecolor='w',markersize=14)
plt.xlim([-1,2])
plt.xticks([0,1],labels=['Group 1','Group 2'])
plt.show()
```

```python
# t-test via stats package

# _ind = independent samples
t1,p1 = stats.ttest_ind(data1,data2)
print(t1)
print(p1)

t2,p2 = stats.ttest_ind(data2,data1)
print(t2)
print(p2)
```

```python
# common way to show t-test results in a plot
fig = plt.figure(figsize=(10,4))
plt.rcParams.update({'font.size':12}) # change the font size

plt.plot(0+np.random.randn(n1)/15, data1, 'ro',markerfacecolor='w',markersize=14)
plt.plot(1+np.random.randn(n2)/15, data2, 'bs',markerfacecolor='w',markersize=14)
plt.xlim([-1,2])
plt.xticks([0,1],labels=['Group 1','Group 2'])

# set the title to include the t-value and p-value
plt.title(f't = {t:.2f}, p={p:.3f}')

plt.show()

# 'ro' and 'bs' (the format strings)
# This is a shorthand for color and marker style:
# 'r' = red
# 'o' = circle marker
# 'b' = blue
# 's' = square marker
# So:
# Group 1 → red circles
# Group 2 → blue squares

# markerfacecolor='w'
# This sets the inside color of the marker to white.
# So the circles and squares are outlined red/blue but hollow inside.

# markersize=14
# This makes the markers bigger (default is ~6).
# 14 = quite large, easier to see individual data points.
```

Normally, if you plotted 0, data1 and 1, data2, you’d get all points stacked in a vertical line at x=0 and x=1. That can make it hard to see how many points overlap.

The trick:
np.random.randn(n1)/15 adds a tiny horizontal jitter.
It’s basically adding a small random “wiggle” around 0 or 1.
Dividing by 15 makes the wiggle small enough so points don’t spread too far away from the group.

So instead of all points sitting exactly at x=0 or x=1, they’re scattered around those positions.

```python
# Create figure
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ---- Plot 1: No jitter ----
axes[0].plot(np.zeros(n1), data1, 'ro', markerfacecolor='w', markersize=10)
axes[0].plot(np.ones(n2),  data2, 'bs', markerfacecolor='w', markersize=10)
axes[0].set_title("No Jitter")
axes[0].set_xlim([-0.5, 1.5])
axes[0].set_xticks([0,1])
axes[0].set_xticklabels(['Group 1','Group 2'])
axes[0].set_ylabel('Values')

# ---- Plot 2: With jitter ----
axes[1].plot(0 + np.random.randn(n1)/15, data1, 'ro', markerfacecolor='w', markersize=10)
axes[1].plot(1 + np.random.randn(n2)/15, data2, 'bs', markerfacecolor='w', markersize=10)
axes[1].set_title("With Jitter")
axes[1].set_xlim([-0.5, 1.5])
axes[1].set_xticks([0,1])
axes[1].set_xticklabels(['Group 1','Group 2'])
axes[1].set_ylabel('Values')
```
